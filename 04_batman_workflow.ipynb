```python
from langgraph.graph import StateGraph, START, END 
from typing import TypedDict
```

```python
class Batesman(TypedDict):
    runs: int
    balles: int
    four: int
    six: int 

    sr: float
    bpb: float
    boundary_percent: float
    summary : str
    
```

```python
def calculate_sr(state: Batesman):
    sr = (state["runs"]/state["balles"])*100
    # state["sr"] = sr
    return {"sr": sr}
```

```python
def calculate_bpb(state: Batesman):
    bpb =  state["balles"]/(state["four"]+ state["six"])
    # state["bpb"] =  bpb
    return {"bpb": bpb}
```

```python
def calculate_boundry_pr(state: Batesman):
    boundary_percent = (((state["four"]*4) + (state["six"]*6))/state["runs"])*100
    # state["boundary_percent"] = bp
    return {"boundary_percent": boundary_percent}
```

```python
def summary(state: Batesman):
    summary =  f"""
    strik Rate - {state['sr']} \n
    Balls per boundary - {state['bpb']} \n
    Boundary parcent -  {state['boundary_percent']}
    """
    # state['summary'] = summary
    return {"summary": summary}
```

```python
graph =  StateGraph(Batesman)

graph.add_node("calculate_sr", calculate_sr)
graph.add_node("calculate_bpb", calculate_bpb)
graph.add_node("calculate_boundry_pr", calculate_boundry_pr)
graph.add_node("summary", summary)

#edges

graph.add_edge(START, 'calculate_sr')
graph.add_edge(START, 'calculate_bpb')
graph.add_edge(START, 'calculate_boundry_pr')


graph.add_edge('calculate_sr', 'summary')
graph.add_edge('calculate_bpb', 'summary')
graph.add_edge('calculate_boundry_pr', 'summary')

graph.add_edge("summary", END)


wrokflow =  graph.compile()
```

```python
initial_state = {
    "runs": 100,
    "balles": 50,
    "four": 6,
    "six" : 4
}

wrokflow.invoke(initial_state)
```
