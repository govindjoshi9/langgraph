```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatOpenAI()
```

```python
class LLMStates(TypedDict):
    question: str
    answer : str
```

```python
def llm_qa(state: LLMStates)->LLMStates:
    # question extraction
    question =  state['question']

    #prompt creation
    promt =  f"Answer this particuler question {question}"
    #ans extraction
    answer = model.invoke(promt)
    #update anse and state
    state['answer'] =  answer

    return state


```

```python
#create our graph

graph = StateGraph(LLMStates)


#state node
graph.add_node("llm_qa", llm_qa)

#add edges
graph.add_edge(START, "llm_qa")
graph.add_edge( "llm_qa", END)

workflow = graph.compile()
```

```python
initial_state = {'question': "What is the capital of France?"}

final_state =  workflow.invoke(initial_state)

print(final_state["answer"])
```
